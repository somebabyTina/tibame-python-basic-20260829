```python
#bmi:weight(kg)/height(m) ^ 2
weight = float(input("輸入體重:"))
height = float(input("輸入身高:"))

print('你的身高是: ',height,'公分')
print('你的體重是: ',weight,'公斤')
print(weight/(height/100)**2)
```

```python
# 1. 預設的 float (浮點數) 型態：會出現微小誤差
print(0.1 + 0.2)
# 執行結果會是：0.30000000000000004 (而不是完美的 0.3)

# 2. Decimal 型態：強制字串轉換為精確的十進位運算
from decimal import Decimal
print(Decimal('0.1') + Decimal('0.2'))
# 執行結果會是：0.3
```

```python
#石頭>剪刀>布
import random

trans = ["剪刀","石頭","布"]

my = int(input("請出拳 0:剪刀、1:石頭、2:布:"))
com = random.randint(0,2)

print("我出:",trans[my])
print("電腦出:",trans[com])

if my == (com +1) % 3 :
    print("我贏了")
elif com == (my +1) % 3:
    print("我輸了")
else:
    print("平手")
```

```python
#石頭>剪刀>布
import random

trans = ["剪刀","石頭","布"]

for i in range(5):

    print(" == Round ",i+1,"==")
    my = int(input("請出拳 0:剪刀、1:石頭、2:布:"))
    com = random.randint(0,2)
    print(" 我出:",trans[my])
    print(" 電腦出:",trans[com])

    if my == (com +1) % 3 :
        print(" 我贏了")
    elif com == (my +1) % 3:
        print(" 我輸了")
    else:
        print(" 平手")
```
